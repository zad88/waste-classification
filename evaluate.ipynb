```python
from pathlib import Path
import numpy as np
import tensorflow as tf

from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix
)

from tensorflow.keras.applications.mobilenet_v2 import preprocess_input

PROJECT_DIR = Path.cwd()
MODEL_PATH = PROJECT_DIR / "models" / "mobilenetv2_best.keras"

IMAGE_SIZE = (224, 224)

CLASS_NAMES = [
    "Cardboard", "Food Organics", "Glass", "Metal",
    "Miscellaneous Trash", "Paper", "Plastic",
    "Textile Trash", "Vegetation"
]

TEST_IMAGES = {
    "metal_test.png": "Metal",
    "plastic_hard_test.png": "Plastic"
}
```

```python
#Load the model.
model = tf.keras.models.load_model(MODEL_PATH)

print("MobileNetV2 model loaded successfully!")
```

```python
#Evaluate your images
actual_labels = []
predicted_labels = []

for filename, actual_category in TEST_IMAGES.items():

    image_path = PROJECT_DIR / filename

    image = tf.keras.utils.load_img(
        image_path,
        target_size=IMAGE_SIZE
    )

    image_array = tf.keras.utils.img_to_array(image)
    image_array = preprocess_input(image_array)
    image_array = np.expand_dims(image_array, axis=0)

    probabilities = model.predict(image_array, verbose=0)[0]

    predicted_index = np.argmax(probabilities)
    predicted_category = CLASS_NAMES[predicted_index]

    confidence = probabilities[predicted_index] * 100

    actual_labels.append(CLASS_NAMES.index(actual_category))
    predicted_labels.append(predicted_index)

    print(f"\nImage: {filename}")
    print(f"Actual: {actual_category}")
    print(f"Predicted: {predicted_category}")
    print(f"Confidence: {confidence:.2f}%")
    print("Result:", "CORRECT" if actual_category == predicted_category else "INCORRECT")
```
